# **Q9:** User se ek string input lein (e.g., 'Python') aur loop ka use karke us string ko reverse print karein.

````python
# type your code here
user_string = input("Ek string enter karein (e.g., 'Python'): ")
reversed_string = ""
for char in user_string:
    reversed_string = char + reversed_string

print("Reverse ki gai string:", reversed_string)
````

# **Q12:** User se ek number lein aur check karein ke wo **Prime Number** hai ya nahi (using loops).

````python
# type your code here
num = int(input("Ek number darj karein: "))
if num > 1:
    for i in range(2, num):
        if (num % i) == 0:
            print(f"{num} ek prime number nahi hai.")
            print(f"Kyunke {i} isko mukammal taqseem (divide) karta hai.")
            break
    else:
        print(f"{num} ek prime number hai!")
else:
    print(f"{num} ek prime number nahi hai.")
````

# **Q16:** User se ek number input lein aur us ka **Factorial** find karein (e.g., 5! = 5*4*3*2*1).

````python
# type your code here
num = int(input("Number darj karein: "))
factorial = 1

if num < 0:
    print("Negative number ka factorial nahi nikal sakta.")
elif num == 0:
    print("0 ka factorial 1 hota hai.")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print(f"{num} ka factorial {factorial} hai.")
````

# **Q18:** Ek manual counting karein ke ek string (e.g., 'Learning Python is fun') mein kitne **Vowels** (a, e, i, o, u) hain.

````python
# type your code here
text = "Learning Python is fun"

vowels = "aeiouAEIOU"

vowel_count = 0

for char in text:
    if char in vowels:
        vowel_count += 1 

print("Total Vowels:", vowel_count)
````

# **Q19:** **Fibonacci Sequence** ke pehle 10 numbers print karein (0, 1, 1, 2, 3, 5, 8...)

````python
# type your code here
n = 10
n1, n2 = 0, 1

print("Fibonacci sequence:")
for i in range(n):
    print(n1, end=" ")
    nth = n1 + n2
    n1 = n2
    n2 = nth
````

# **Q20:** Ek 'Number Guessing Game' banayein jahan program ek number (1-20) fix kare aur user ko 3 chances de guess karne ke liye. `while loop` aur `break` ka use karein.

````python
# type your code here
import random

# Program ek random number 1 se 20 ke beech fix karega
secret_number = random.randint(1, 20)
chances = 3

print("Welcome to the Number Guessing Game!")
print("Main ne 1 se 20 ke beech ek number soch liya hai. Kya aap use 3 chances mein guess kar sakte hain?")

while chances > 0:
    guess = int(input("\nApna guess enter karein: "))
    
    if guess == secret_number:
        print("Sahi Jawab! Aap jeet gaye! 🎉")
        break 
    elif guess < secret_number:
        print("Aapka guess chhota hai. Thoda bada number try karein.")
    else:
        print("Aapka guess bada hai. Thoda chhota number try karein.")
        
    chances -= 1
    print(f"Aapke paas {chances} chances bache hain.")
if chances == 0:
    print(f"\nAfsos, aapke saare chances khatam ho gaye! Sahi number {secret_number} tha.")
````
